```python
import sqlite3
```

```python
#Connect to an SQLite DataBase
connection=sqlite3.connect('hello.db')
connection
```

```python
#Create a Cursor
cursor=connection.cursor()
```

```python
# Create a table
cursor.execute("""
Create Table Employee(
Id Integer Primary Key,
Name Text Not Null ,
Age Integer,
Department Text
)
""")
```

```python
#Select 
cursor.execute('Select * From Employee')
```

```python
#Insert
cursor.execute('''
Insert Into Employee(Name,Age,Department)
Values("Zahid",22,"AI")
''')

cursor.execute('''
Insert Into Employee(Name,Age,Department)
Values("Tayab",23,"ML")
''')

cursor.execute('''
Insert Into Employee(Name,Age,Department)
Values("Khan",20,"Data Science")
''')

cursor.execute('''
Insert Into Employee(Name,Age,Department)
Values("Arosh",21,"GenAI")
''')

cursor.execute('''
Insert Into Employee(Name,Age,Department)
Values("Furkhan",20,"Data Analyst")
''')
```

```python
#Commit
connection.commit()
```

```python
#Query the data from the table itself
cursor.execute('Select * From Employee')
rows=cursor.fetchall()
```
