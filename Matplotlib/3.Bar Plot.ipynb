```python
import matplotlib.pyplot as plt
```

```python
#Product Sales
products = ["Laptop", "Mobile", "Tablet", "Watch", "Headphones"]
sales = [320, 580, 240, 180, 420]
plt.bar(products,sales,color="blue")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
#Programming Language Usage
languages = ["Python", "Java", "C++", "JavaScript", "Go"]
developers = [45, 30, 22, 38, 15]
plt.bar(languages,developers,color="orange")
plt.title("Programming Language Usage")
plt.xlabel("Languages")
plt.ylabel("Developers")
plt.show()
```

```python
# Monthly Expenses
categories = ["Rent", "Food", "Travel", "Education", "Shopping"]
expenses = [12000, 6000, 2500, 4000, 3500]
plt.bar(categories,expenses,color="pink")
plt.title("Monthly Expenses")
plt.xlabel("Categories")
plt.ylabel("Expenses")
plt.show()
```

```python
#ML Model Accuracy
models = ["Logistic Regression", "Decision Tree", "Random Forest",
          "SVM", "Neural Network"]

accuracy = [82, 85, 91, 88, 94]
plt.bar(models,accuracy,color="yellow")
plt.title("ML Model Accuracy")
plt.xlabel("Models")
plt.ylabel("Accuracy")
plt.show()
```

```python
#Company Department Employees
departments = ["Engineering", "HR", "Sales", "Marketing", "Finance"]
employees = [120, 25, 80, 45, 35]
plt.bar(departments,employees,color="red")
plt.title("Company Department Employees")
plt.xlabel("Departments")
plt.ylabel("Employees")
plt.show()
```
