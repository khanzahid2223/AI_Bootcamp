```python
import matplotlib.pyplot as plt 
```

```python
#Create 2 subplots:
# Line plot of sales
# Bar plot of sales
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales1 = [120, 150, 180, 140, 200, 250, 220]
sales2 = [110, 111, 190, 180, 100, 220, 290]
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.plot(days,sales1,color="green")

plt.subplot(2,2,4)
plt.bar(days,sales2,color="yellow")
plt.show()
```

```python
#2. Create 2×2 subplots containing:
# Line plot
# Bar plot
# Histogram
# Scatter plot
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [120, 150, 180, 140, 200, 250, 220]
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.plot(days,sales,color="pink")

plt.subplot(2,2,2)
plt.bar(days,sales,color="red")

plt.subplot(2,2,3)
plt.hist(sales,bins=5,color="green",edgecolor="black")

plt.subplot(2,2,4)
plt.scatter(days,sales,color="orange",marker="o")
plt.show()
```

```python
#4. Create subplots showing a student's:
# Math marks
# Science marks
# English marks
# Computer Science marks
Names=["Zahid","Tayab","Arosh","Furkhan","Pathan"]
Maths=[99,90,67,54,81]
Science=[100,87,45,67,99]
Eng=[98,78,56,34,87]
Computer=[99,100,98,81,92]
plt.figure(figsize=(15,15))
plt.subplot(2,2,1)
plt.bar(Names,Maths,color="red")
plt.title("Student scored marks in Maths")
plt.xlabel("Name of the Students")
plt.ylabel("Marks scored by students")

plt.subplot(2,2,2)
plt.bar(Names,Science,color="pink")
plt.title("Student scored marks in Science")
plt.xlabel("Name of the Students")
plt.ylabel("Marks scored by students")

plt.subplot(2,2,3)
plt.bar(Names,Eng,color="blue")
plt.title("Student scored marks in English")
plt.xlabel("Name of the Students")
plt.ylabel("Marks scored by students")

plt.subplot(2,2,4)
plt.bar(Names,Computer,color="green")
plt.title("Student scored marks in Computer Science")
plt.xlabel("Name of the Students")
plt.ylabel("Marks scored by students")

plt.show()
```

```python
# Company Sales Dashboard
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 140, 135, 160, 180, 210]
profit = [20, 25, 22, 30, 35, 45]
# → Create 2 subplots:
# 1 → Line plot of sales
# 2 → Bar plot of profit
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.plot(months,sales,color="red",marker=">",linestyle="--")
plt.title("Company Sales Dashboard")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.subplot(2,2,4)
plt.bar(months,profit,color="purple")
plt.title("Company Sales Dashboard")
plt.xlabel("Months")
plt.ylabel("Profit")
plt.show()
```

```python
#Student Performance Dashboard
subjects = ["Math", "Science", "English", "Python"]
marks = [88, 82, 91, 95]
attendance = [92, 85, 90, 96]
# → Create 2 subplots:
# 1 → Bar plot of marks
# 2 → Bar plot of attendance
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.bar(subjects,marks,color="pink")
plt.title("Student Performance Dashboard")
plt.xlabel("Subjects")
plt.ylabel("Marks of the Student")

plt.subplot(2,2,2)
plt.bar(subjects,attendance,color="yellow")
plt.title("Student Performance Dashboard")
plt.xlabel("Subjects")
plt.ylabel("Attendance of the Student")
plt.show()
```

```python
#E-commerce Dashboard
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
orders = [120, 150, 135, 180, 220, 300, 270]
revenue = [24000, 30000, 27000, 36000, 44000, 60000, 54000]
# → Create 2 subplots:
# 1 → Line plot of orders
# 2 → Line plot of revenue
plt.figure(figsize=(20,15))
plt.subplot(2,2,1)
plt.plot(days,orders,color="blue",marker="s",linestyle=":")
plt.title("E-commerce Dashboard")
plt.xlabel("Days")
plt.ylabel("Orders")

plt.subplot(2,2,2)
plt.plot(days,revenue,color="purple",marker="d",linestyle="--")
plt.title("E-commerce Dashboard")
plt.xlabel("Days")
plt.ylabel("Revenue")
plt.show()
```

```python
#Machine Learning Dashboard
epochs = [1, 2, 3, 4, 5, 6, 7, 8]
accuracy = [65, 72, 78, 82, 86, 89, 92, 94]
loss = [0.90, 0.75, 0.62, 0.51, 0.43, 0.37, 0.32, 0.28]
# → Create 2 subplots:
# 1 → Line plot of accurac
# 2 → Line plot of loss
plt.figure(figsize=(20,15))
plt.subplot(2,2,1)
plt.grid(True)
plt.plot(epochs,accuracy,color="red",marker="d",linestyle="--")
plt.title("Machine Learning Dashboard")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")

plt.subplot(2,2,2)
plt.grid(True)
plt.plot(epochs,loss,color="orange",marker="s",linestyle=":")
plt.title("Machine Learning Dashboard")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()
```

```python
#Complete Data Analysis Dashboard
products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [120, 250, 150, 100]
prices = [60000, 25000, 30000, 8000]
ratings = [4.5, 4.2, 4.0, 4.4]
plt.figure(figsize=(15,10))
plt.title("Complete Data Analysis Dashboard")
plt.subplot(2,2,1)
plt.plot(products,sales,color="red",marker="o",linestyle="--")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.subplot(2,2,2)
plt.plot(products,prices,color="green",marker="^",linestyle="-.")
plt.xlabel("Products")
plt.ylabel("Prices")

plt.subplot(2,2,3)
plt.plot(products,ratings,color="purple",marker=">",linestyle=":")
plt.xlabel("Products")
plt.ylabel("Ratings")
plt.show()
```
