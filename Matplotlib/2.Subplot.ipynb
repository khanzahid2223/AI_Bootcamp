```python
import matplotlib.pyplot as plt 
```

```python
#Create 2 subplots:
# Line plot of sales
# Bar plot of sales
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales1 = [120, 150, 180, 140, 200, 250, 220]
sales2 = [110, 111, 190, 180, 100, 220, 290]
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.plot(days,sales1,color="green")

plt.subplot(2,2,4)
plt.bar(days,sales2,color="yellow")
plt.show()
```

```python
#2. Create 2×2 subplots containing:
# Line plot
# Bar plot
# Histogram
# Scatter plot
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [120, 150, 180, 140, 200, 250, 220]
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.plot(days,sales,color="pink")

plt.subplot(2,2,2)
plt.bar(days,sales,color="red")

plt.subplot(2,2,3)
plt.hist(sales,bins=5,color="green",edgecolor="black")

plt.subplot(2,2,4)
plt.scatter(days,sales,color="orange",marker="o")
plt.show()
```

```python
#4. Create subplots showing a student's:
# Math marks
# Science marks
# English marks
# Computer Science marks
Names=["Zahid","Tayab","Arosh","Furkhan","Pathan"]
Maths=[99,90,67,54,81]
Science=[100,87,45,67,99]
Eng=[98,78,56,34,87]
Computer=[99,100,98,81,92]
plt.figure(figsize=(15,15))
plt.subplot(2,2,1)
plt.bar(Names,Maths,color="red")
plt.title("Student scored marks in Maths")
plt.xlabel("Name of the Students")
plt.ylabel("Marks scored by students")

plt.subplot(2,2,2)
plt.bar(Names,Science,color="pink")
plt.title("Student scored marks in Science")
plt.xlabel("Name of the Students")
plt.ylabel("Marks scored by students")

plt.subplot(2,2,3)
plt.bar(Names,Eng,color="blue")
plt.title("Student scored marks in English")
plt.xlabel("Name of the Students")
plt.ylabel("Marks scored by students")

plt.subplot(2,2,4)
plt.bar(Names,Computer,color="green")
plt.title("Student scored marks in Computer Science")
plt.xlabel("Name of the Students")
plt.ylabel("Marks scored by students")

plt.show()
```
