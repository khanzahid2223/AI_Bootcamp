```python
import matplotlib.pyplot as plt
#Monthly Budget
categories = ["Rent", "Food", "Travel", "Education", "Other"]
amount = [12000, 6000, 2500, 4000, 2500]
# ex=(0.1,0.1,0,0.1,0)
plt.pie(amount,labels=categories,autopct="%1.1f%%")
plt.title("Monthly Budget",color="red")
plt.show()
```

```python
# Smartphone Market
brands = ["Samsung", "Apple", "Xiaomi", "OnePlus", "Others"]
market_share = [28, 25, 20, 12, 15]
plt.pie(market_share,labels=brands,autopct="%1.1f%%")
plt.title(" Smartphone Market")
plt.show()
```

```python
#E-commerce Sales
categories = ["Electronics", "Clothing", "Groceries", "Books", "Others"]
sales = [35, 25, 20, 10, 10]
plt.pie(sales,labels=categories,autopct="%1.1f%%")
plt.title("E-commerce Sales")
plt.show()
```

```python
#ML Dataset Split
dataset = ["Training", "Validation", "Testing"]
percentage = [70, 15, 15]
plt.pie(percentage,labels=dataset,autopct="%1.1f%%")
plt.title("ML Dataset Split")
plt.show()
```

```python
# Website Traffic Sources
sources = ["Google", "YouTube", "Instagram", "Direct", "Other"]
traffic = [40, 25, 15, 12, 8]
plt.pie(traffic,labels=sources,autopct="%1.1f%%")
plt.title("Website Traffic Sources")
plt.show()
```
