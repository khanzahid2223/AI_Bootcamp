```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [120, 150, 180, 140, 200, 250, 220]
# 1. Create a line plot showing sales for each day.
plt.plot(days,sales)
plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [120, 150, 180, 140, 200, 250, 220]
#2. Add a suitable title, x-axis label, and y-axis label.
plt.plot(days,sales)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales of each day")
plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [120, 150, 180, 140, 200, 250, 220]
#3. Add markers to every data point.
plt.plot(days,sales,marker="o")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales of each day")
plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [120, 150, 180, 140, 200, 250, 220]
#4. Change the line style to dashed.
plt.plot(days,sales,marker="o",linestyle="--")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales of each day")
plt.grid()
plt.show()
```
