```python
import seaborn as sns
```

```python
df=sns.load_dataset('titanic')
df.head()
```

```python
#Check missing Values
df.isnull().sum()
```

```python
#Delete the datapoints to handle missing values
# df.dropna().shape
```

```python
#Column wise
df.dropna(axis=1)
```

```python
#Implutation Techniques
#1 Mean Value Implication
df['Age_Mean']=df['age'].fillna(df['age'].mean())
df[['Age_Mean','age']]
```

```python
#2 Median Value Implication
df['Age_Median']=df['age'].fillna(df['age'].median())
df[['Age_Median','age']]
```

```python
#3 Mode value Implication
df[df['embarked'].isnull()]
mv=df[df['embarked'].notna()]['embarked'].mode()[0]
df['eb']=df['embarked'].fillna(mv)
df[['eb','embarked']]
```
