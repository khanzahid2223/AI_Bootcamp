```python
import numpy as np
```

```python
lst_marks=[45,32,56,75,89,54,32,89,96,87,54,45,98,99,67,74]
minimum,Q1,Median,Q3,Maximum=np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
print(minimum,Q1,Median,Q3,Maximum)
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
#Low Fence
low_fence=Q1-1.5*(IQR)
print(low_fence)
```

```python
#High Fence
high_fence=Q3+1.5*(IQR)
print(high_fence)
```

```python
import seaborn as sns
#Without Outliers
sns.boxplot(lst_marks)
```

```python
#With Outliers
lst_marks=[-100,-200,45,32,56,75,89,54,32,89,96,87,54,45,98,99,67,74,150,170,180]
sns.boxplot(lst_marks)
```
